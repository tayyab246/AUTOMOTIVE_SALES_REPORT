```python
import pandas as pd

# Load the cleaned data
df = pd.read_csv('cleaned_car_sales.csv')

# 1. Total Market Sales by Manufacturer
sales_by_brand = df.groupby('Manufacturer')['Sales'].sum().sort_values(ascending=False)

# 2. Average Depreciation (Price minus Resale Value)
df['Depreciation'] = df['Price'] - df['Resale_Value']
avg_depreciation = df.groupby('Vehicle_type')['Depreciation'].mean()

# 3. Correlation between Price and Horsepower
correlation = df['Price'].corr(df['Horsepower'])

print("--- Analysis Highlights ---")
print("\nTop 5 Brands by Total Sales Volume:")
print(sales_by_brand.head(5).apply(lambda x: f"{x:,.0f} units"))

print("\nAverage Depreciation by Vehicle Type:")
print(avg_depreciation.apply(lambda x: f"${x:,.2f}"))

print(f"\nCorrelation between Price and Horsepower: {correlation:.2f}")
```
